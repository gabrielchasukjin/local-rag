```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
from langchain.llms import HuggingFacePipeline
import torch
```

```python
# a smaller model suitable for M3 Mac
model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.float16,  # Use float16 for M3
    device_map="auto"
)
pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=256,
    do_sample=True,
    temperature=0.25,  # Low temperature for more focused answers
    repetition_penalty=1.2
)
llm = HuggingFacePipeline(pipeline=pipe)
```

```python
def chat_with_llm(llm, message, history=None):
    # Initialize history if None
    if history is None:
        history = []
    
    # Format the conversation history and current message
    if not history:
        prompt = f"<|user|>\n{message}\n<|assistant|>\n"
    else:
        prompt = ""
        for h in history:
            prompt += f"<|user|>\n{h['user']}\n<|assistant|>\n{h['assistant']}\n"
        prompt += f"<|user|>\n{message}\n<|assistant|>\n"
    
    # Generate response
    response = llm.invoke(prompt).strip()
    
    # Update history
    history.append({
        "user": message,
        "assistant": response
    })
    
    return response, history
```

```python
import warnings
warnings.filterwarnings("ignore", message="To copy construct from a tensor")
```

```python
def interactive_chat(llm):
    history = []
    print("Chat with TinyLlama (type 'exit' to quit)")
    
    while True:
        user_input = input("\nYou: ")
        
        if user_input.lower() == 'exit':
            print("Goodbye!")
            break
        
        response, history = chat_with_llm(llm, user_input, history)
        print(f"\nTinyLlama: {response}")

# Start the interactive chat
interactive_chat(llm)
```
